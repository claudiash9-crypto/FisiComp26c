```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
#from matplotlib import animation
from IPython.display import HTML 
#no necesaria para la animacion pero si para jupyter
```

```python
%%capture
fig , ejes = plt.subplots()
x = np.linspace(0,2*np.pi)

linea, = ejes.plot(x,np.cos(x))

def cuadro(n):
    linea.set_ydata( np.cos(x - n/20) )
    return linea,

animacion = FuncAnimation(fig,cuadro,frames=100, interval = 45,blit=True)
```

```python
HTML(animacion.to_html5_video())
```

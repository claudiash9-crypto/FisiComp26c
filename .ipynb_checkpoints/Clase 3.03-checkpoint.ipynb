```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def deriv_sim(f,x):
    df = np.zeros_like(f)
    for i in range(1,len(x)-1):
        df[i] = (f[i+1]-f[i-1])/(x[i+1]-x[i-1])
    df[0] = 0.5*(-3*f[0] + 4*f[1]- f[2])/(x[1]-x[0])
    df[-1] = 0.5*(3*f[-1]-4*f[-2] + f[-2])/(x[-1]-x[-2])
    return df
```

consideremos un potencial  
      $V= \frac{1}{r^4} - \frac{2}{r^2}$

```python
h = 0.001
x = np.arange(1e-3, 5+h, h)
y= 1/x**4 -2/x**2
plt.plot(x,y)
plt.ylim([-2,5])
plt.grid()
plt.show()
```

```python
dy = -4/x**5 + 4/x**3
dy_num= deriv_sim(y,x)
plt.plot(x,dy, label=r"f' analítica")
plt.plot(x,dy_num,'--', label=r"f' numérica")
plt.ylim([-2,1])
plt.grid()
plt.show()
```

```python
error = abs(dy-dy_num)
plt.semilogy(x,error)
# crea un gráfico 2D con escala lineal en el eje x 
#y escala logarítmica (base 10) en el eje y. 
#Es ideal para visualizar datos que cubren varios órdenes de magnitud en el eje vertical, como funciones exponenciales,
#plt.ylim([0,1])
plt.grid()
plt.show()
```

# ***Resolviendo el problema por partes***

```python
h1 = 1e-4
pt_empalme = 1
x1 = np.arange(0.1,pt_empalme,h1)
h2 = 1e-3
x2 = np.arange(pt_empalme,5,h2)
x= np.concatenate((x1,x2))
print(len(x1),len(x2), len(x))
y= 1/x**4 -2/x**2
plt.plot(x,y)
plt.ylim([-2,5])
plt.grid()
plt.show()
```

```python
z1 = x**4
z2 = x**2
z_y =1/x
dy = -4/x**5 +4/x**3
dz1_num, dz2_num = deriv_sim(z1,x), deriv_sim(z2,x)
dzdx = deriv_sim(z_y,x)
dy_alt= (16/dz1_num - 8/dz2_num)*dzdx
plt.plot(x,dy, label= r"f' (analítica)")
plt.plot(x,dy_alt, '--', label= r"f'(alt)")
plt.ylim([-2,1])
plt.grid(), plt.legend()
plt.show()
```

```python
error_alt= abs(dy-dy_alt)
plt.semilogy(x,error_alt,label= 'alternativo')
plt.grid(), plt.legend()
plt.show()
```

```python
import numpy as np
import random as rd
import matplotlib.pyplot as plt
```

```python
def M_Jac_iter(V,V0=1, err=1e-3, step=20):
    diferencia = 1
    while diferencia > err:
        for i in range(step):
            V = M_Jac_iter(V,V0=V0)
        V_ref = M_Jac_iter(np.copy(V))
        diferencia = np.max(abs(V_ref-V))
    return V

def V_cuadrado(x,y,L=0.5, V0=1.0):
    Nx, Ny = len(x), len(y)
    malla = np.zeros((Nx,Ny))
    for i in range(1,Nx-1):
        for j in range(1,Ny-1):
            if -0.25*L < x[i] < 0.5*L and 0.5*L < y[j] < 0.5*L:
                malla[i,j] = 1
            else:
                malla[i,j] = rd.random()
    return V0*malla

def V_front_cuad(x,y,electrodo):
    Nx,Ny = len(x), len(y)
    malla = np.zeros((Nx,Ny))
    for i in range(1,Nx-1):
        for j in range(1,Ny-1):
            if electrodo(x[i],y[j]) == True:
                malla[i,j] = 1
            else:
                malla[i,j] = rd.random()
    return V0*malla

def M_mouse(x,y):
    respuesta = False
    c1 = x**2 +y**2
    c2, c3 =(x-0.75)**2 + (y-0.75)**2, (x+0.75)**2 + (y-0.75)**2
    if c1 <= 1 or c2 <= 0.5 or c3 <= 0.5:
        respuesta = True
    return respuesta
    
```

```python
x,y = np.linspace(-2,2,100), np.linspace(-2,2,100)
V = V_front_cuad(x,y,M_mouse)
plt.inshow(V.transpose(), 
           extent=(x[0],x[-1],y[0],y[-1]),
           origin='lower',
           cmap='bone')
plt.show()
```
